# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

The above graph makes it more distinct to understand the difference between the two weeks. It feels obvious that the earlier week is on the left.

The later graph has no obvious visual distinction between the two, making it harder to easily digest the data. Overall, I prefer the first to quickly communicate trends on a weekly basis.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Product A': [390, 410, 430, 400, 320, 300, 280],
    'Product B': [370, 400, 420, 390, 310, 290, 275],
    'Product C': [90, 100, 110, 180, 200, 210, 205],
    'Product D': [100, 120, 130, 200, 250, 230, 225],
    'Product E': [50, 60, 75, 100, 180, 200, 190]
}
years = [2008, 2009, 2010, 2011, 2012, 2013, 2014]

df = pd.DataFrame(data, index=years)

plt.figure(figsize=(10, 6))
for product in df.columns:
    plt.plot(df.index, df[product], label=product, linewidth=2)

plt.title("Average Retail Product Price per Year")
plt.xlabel("Year")
plt.ylabel("Price ($)")
plt.ylim(0, 500)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(title="Product", loc="lower left")
plt.tight_layout()
plt.show()
```
